```python
import numpy as np 
import pandas as pd 
from sklearn.preprocessing import MinMaxScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, mean_absolute_percentage_error, r2_score
```

```python
df = pd.read_csv('Data/train.csv')
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.dropna(subset=['x', 'y'], inplace=True)
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
scaler_x = MinMaxScaler()
scaler_y = MinMaxScaler()

x = df['x']
y = df['y']

x_scaled = scaler_x.fit_transform(np.array(x).reshape(-1, 1))
y_scaled = scaler_y.fit_transform(np.array(y).reshape(-1, 1))
```

```python
regression = LinearRegression()

model = regression.fit(x_scaled, y_scaled)
```

```python
test_df = pd.read_csv('Data/test.csv')
```

```python
test_df.head()
```

```python
x_test = test_df['x']
y_test = test_df['y']

x_test_scaled = scaler_x.fit_transform(np.array(x_test).reshape(-1, 1))
```

```python
y_pred = model.predict(x_test_scaled)
```

```python
y_pred_inversed = scaler_y.inverse_transform(y_pred)
```

```python
print(mean_absolute_error(y_test, y_pred_inversed))
print(mean_squared_error(y_test, y_pred_inversed))
print(mean_absolute_percentage_error(y_test, y_pred_inversed) * 100)
print(r2_score(y_test, y_pred_inversed))
```
